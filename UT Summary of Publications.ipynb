## UT Knoxville

```python
# Read in the full data from ORIED

utKnox = pd.read_excel('UTK Publication Data.xlsx') #Contains all information from 2014-2023
```

```python
# This data was gathered in October 2024, and thus is missing some 2024 data. As a result, only 2014-2023 will be gathered from this file.

utKnox['Publication Year'] = utKnox['Publication Year'].astype(int)

utKnox1423 = utKnox[(utKnox['Publication Year'] >= 2014) & (utKnox['Publication Year'] <= 2023)]

utKnox1423
```

### Add in 2024 and 2025 Data from WoS

```python
utKnox24 = pd.read_excel('UTK WoS 2024.xlsx')

utKnox24 = utKnox24[utKnox24['Publication Year'] <= 2024]
```

```python
utKnox25 = pd.read_excel('UTK WoS 2025.xlsx')

utKnox25 = utKnox25[utKnox25['Publication Year'] <= 2025]
```

```python
utKnox = pd.concat([utKnox1423, utKnox24, utKnox25])

utKnox['Publication Campus'] = 'UT Knoxville'
```

### Combine Publishers with Multiple Names in the Data

If there are any other publisher acquisitions/mergers or misnames not already caught in the code chunk below, add them in with a new "elif" statement.

```python
utKnox['Unified Publisher'] = ''

for i in range(len(utKnox['Publication Campus'])):
    inputValue = utKnox['Publisher'].iloc[i]
    if re.match('.*ELSEVIER.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*CHURCHILL.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*CHURCHHILL.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*SAUNDERS.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*CELL PRESS.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*NATURE.*', inputValue):
        unifiedPublisher = 'Nature'
    elif re.match('.*IOP.*', inputValue):
        unifiedPublisher = 'IOP'
    elif re.match('.*SAGE.*', inputValue):
        unifiedPublisher = 'Sage'
    elif re.match('.*WILEY.*', inputValue):
        unifiedPublisher = 'Wiley'
    elif re.match('.*HINDAWI.*', inputValue):
        unifiedPublisher = 'Wiley'
    elif re.match('.*OXFORD.*', inputValue): 
        unifiedPublisher = 'OUP'
    elif re.match('.*SPRINGER.*', inputValue):
        unifiedPublisher = 'Springer'
    elif re.match('.*BMC.*', inputValue):
        unifiedPublisher = 'Springer'
    elif re.match('.*NATURE.*', inputValue):
        unifiedPublisher = 'Nature'
    elif re.match('.*MDPI.*', inputValue):
        unifiedPublisher = 'MDPI'
    elif re.match('.*IEEE.*', inputValue):
        unifiedPublisher = 'IEEE'
    elif re.match('.*CAMBRIDGE.*', inputValue):
        unifiedPublisher = 'Cambridge'
    elif re.match('.*TAYLOR & FRANCIS.*', inputValue):
        unifiedPublisher = 'Taylor & Francis'
    else:
        unifiedPublisher = inputValue.title()
    utKnox['Unified Publisher'].iloc[i] = unifiedPublisher
```

### Select Relevant Columns and Save to Database

```python
utKnoxClean = utKnox[['Publication Type', 'Authors', 'Book Authors', 'Book Editors', 'Book Group Authors', 'Author Full Names', 'Book Author Full Names', 'Group Authors', 'Article Title', 'Source Title', 'Book Series Title', 'Document Type', 
                      'Conference Title', 'Conference Sponsor', 'Author Keywords', 'Abstract', 'Addresses', 'Affiliations', 'ORCIDs', 'Funding Orgs', 'Funding Name Preferred', 'Funding Text', 'Cited Reference Count', 'Times Cited, WoS Core', 'Times Cited, All Databases',
                      '180 Day Usage Count', 'Since 2013 Usage Count', 'Publisher', 'Publisher City', 'Publisher Address', 'ISSN', 'eISSN', 'ISBN', 'Journal Abbreviation', 'Publication Date', 'Publication Year', 'Volume', 'Issue', 'Start Page', 'End Page', 'DOI', 'DOI Link', 'Book DOI', 
                      'Number of Pages', 'WoS Categories', 'Research Areas', 'Open Access Designations', 'Highly Cited Status', 'Hot Paper Status', 'Date of Export', 'Web of Science Record', 'Publication Campus', 'Unified Publisher']]

utKnoxClean = utKnoxClean.astype('str')
```
