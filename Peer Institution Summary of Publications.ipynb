## All campuses listed separately are included: Auburn, Clemson, Iowa State, Colorado-Boulder, Kentucky, Missouri, Nebraska-Lincoln, Oklahoma, South Carolina, and Virginia Tech.

## The data ranges from 2014-2024 and includes all Article publications.

```python
import pandas as pd
import sqlite3
import warnings
import re
import numpy as np
import math
import os
from os import listdir
from os.path import isfile, join
warnings.filterwarnings("ignore")
```

## Connect to Database

```python
conn = sqlite3.connect(r"C:\Users\aharrima\OneDrive - University of Tennessee\Documents\SEC_and_Peer_Publications.db")
cur = conn.cursor()
```

## Auburn

```python
auburnWOS = pd.read_excel('Auburn WoS 2014-2024.xlsx')
```

```python
auburnWOS = auburnWOS.loc[:, ~auburnWOS.columns.str.contains('^Unnamed')]
```

### Combine Publishers with Multiple Names in the Data

If there are any other publisher acquisitions/mergers or misnames not already caught in the code chunk below, add them in with a new "elif" statement.

```python
auburnWOS['Publication Campus'] = 'Auburn University'

auburnWOS['Unified Publisher'] = ''

for i in range(len(auburnWOS['Publication Campus'])):
    inputValue = auburnWOS['Publisher'].iloc[i]
    if isinstance(inputValue, float) == True:
        unifiedPublisher = 'NA'
    elif re.match('.*Elsevier.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*Churchill.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*Churchhill.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*Saunders.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*Cell Press.*', inputValue):
        unifiedPublisher = 'Elsevier'
    elif re.match('.*Nature.*', inputValue):
        unifiedPublisher = 'Nature'
    elif re.match('.*IOP.*', inputValue):
        unifiedPublisher = 'IOP'
    elif re.match('.*SAGE.*', inputValue):
        unifiedPublisher = 'Sage'
    elif re.match('.*Wiley.*', inputValue):
        unifiedPublisher = 'Wiley'
    elif re.match('.*Hindawi.*', inputValue):
        unifiedPublisher = 'Wiley'
    elif re.match('.*Oxford.*', inputValue): 
        unifiedPublisher = 'OUP'
    elif re.match('.*Springer.*', inputValue):
        unifiedPublisher = 'Springer'
    elif re.match('.*BMC.*', inputValue):
        unifiedPublisher = 'Springer'
    elif re.match('.*MDPI.*', inputValue):
        unifiedPublisher = 'MDPI'
    elif re.match('.*Institute of Electrical and Electronics Engineers.*', inputValue):
        unifiedPublisher = 'IEEE'
    elif re.match('.*Cambridge.*', inputValue):
        unifiedPublisher = 'Cambridge'
    elif re.match('.*Taylor & Francis.*', inputValue):
        unifiedPublisher = 'Taylor & Francis'
    else:
        unifiedPublisher = inputValue.title()
    auburnWOS['Unified Publisher'].iloc[i] = unifiedPublisher
```

### Clean Data and Save to Database

```python
auburnWOS = auburnWOS.astype({'Pubmed Id': object, 'Cited References': object, 
                        'Cited Reference Count': np.int8, 'Times Cited, WoS Core': np.int8, 'Times Cited, All Databases': np.int8, '180 Day Usage Count': np.int8, 'Since 2013 Usage Count': np.int8, 'Number of Pages': np.int8})
```

```python
auburnWOS.to_sql('wos_pubs_auburn', conn, if_exists = 'replace', index = False)
```
